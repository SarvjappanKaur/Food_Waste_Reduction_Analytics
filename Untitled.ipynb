```python
#filling temperature based on seasonality and storage cnditions
df['Temperature'] = df['Temperature'].fillna(
    df.groupby(['Seasonality', 'Storage Conditions'])['Temperature']
      .transform('median')
)
```

```python
#filling humidity based on seasonality and storage cnditions
df['Humidity'] = df['Humidity'].fillna(
    df.groupby(['Seasonality', 'Storage Conditions'])['Humidity']
      .transform('median')
)
```

```python
#deleting customer_count
df.drop('Customer_Count', axis=1, inplace=True)
```

```python
# Handling quantity_sold
df['Guest_Group'] = pd.cut(
    df['Number of Guests'],
    bins=[0, 20, 50, 100, 200, float('inf')],
    labels=['1-20', '21-50', '51-100', '101-200', '200+']
)
```

```python
df['Quantity_Sold'] = df['Quantity_Sold'].fillna(
    df.groupby(
        ['Food_Item', 'Seasonality', 'Guest_Group'],
        observed=True
    )['Quantity_Sold'].transform('median')
)
```
